### Creating visual summaries of the variables in the data set, to search for extreme values and externalities. that might be affecting the summary of the data. 

```python
import matplotlib.pyplot as plt
import seaborn as sns
pd.options.display.float_format = '{:,.0f}'.format
columns = [
    "ClosePrice", "ListPrice", "OriginalListPrice",
    "LivingArea", "LotSizeAcres", "BedroomsTotal",
    "BathroomsTotalInteger", "DaysOnMarket", "YearBuilt", 
]

def analyze_dataset(df, name):
    print(f"\n===== Analysis for {name} =====")
    
    for col in columns:
        print(f"\n--- {col} ---")
        
        # Summary stats
        print(df[col].describe(percentiles=[0.01, 0.05, 0.5, 0.95, 0.99]))
        
        # Outliers (IQR)
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        
        outliers = df[(df[col] < lower) | (df[col] > upper)]
        print(f"Outliers: {len(outliers)}")
        
        # Plots
        plt.figure(figsize=(12,4))
        
        plt.subplot(1,2,1)
        sns.histplot(df[col], kde=True)
        plt.title(f"{name} - {col} Histogram")
        
        plt.subplot(1,2,2)
        sns.boxplot(x=df[col])
        plt.title(f"{name} - {col} Boxplot")
        
        plt.tight_layout()
        plt.show()
```

```python
analyze_dataset(df_sold_new, "Sold Data")
analyze_dataset(df_listing_new, "Listings Data")
```

## Merging Mortgage data. 

```python
fred_url = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=MORTGAGE30US"

mortgage = pd.read_csv(fred_url)

print("Original mortgage data columns:")
print(mortgage.columns)

print("\nMortgage data preview:")
print(mortgage.head())
```

```python
if "DATE" in mortgage.columns:
    mortgage = pd.read_csv(fred_url, parse_dates=["DATE"])
    mortgage.columns = ["date", "rate_30yr_fixed"]
elif "observation_date" in mortgage.columns:
    mortgage = pd.read_csv(fred_url, parse_dates=["observation_date"])
    mortgage.columns = ["date", "rate_30yr_fixed"]
else:
    raise ValueError("Unexpected FRED mortgage data column names.")

mortgage["rate_30yr_fixed"] = pd.to_numeric(
    mortgage["rate_30yr_fixed"],
    errors="coerce"
)

mortgage["year_month"] = mortgage["date"].dt.to_period("M")

mortgage_monthly = (
    mortgage.groupby("year_month")["rate_30yr_fixed"]
    .mean()
    .reset_index()
)

print("Monthly mortgage rate data:")
print(mortgage_monthly.head())
```

```python
df_sold_new["CloseDate"] = pd.to_datetime(df_sold_new["CloseDate"], errors="coerce")
df_listing_new["ListingContractDate"] = pd.to_datetime(df_listing_new["ListingContractDate"], errors="coerce")

df_sold_new["year_month"] = df_sold_new["CloseDate"].dt.to_period("M")
df_listing_new["year_month"] = df_listing_new["ListingContractDate"].dt.to_period("M")

sold_with_rates = df_sold_new.merge(
    mortgage_monthly,
    on="year_month",
    how="left"
)

listed_with_rates = df_listing_new.merge(
    mortgage_monthly,
    on="year_month",
    how="left"
)

print("Missing mortgage rates in sold dataset:", sold_with_rates["rate_30yr_fixed"].isnull().sum())
print("Missing mortgage rates in listed dataset:", listed_with_rates["rate_30yr_fixed"].isnull().sum())
```

```python
print("Sold preview with mortgage rates:")
print(sold_with_rates[["CloseDate", "year_month", "ClosePrice", "rate_30yr_fixed"]].head())

print("\nListed preview with mortgage rates:")
print(listed_with_rates[["ListingContractDate", "year_month", "ListPrice", "rate_30yr_fixed"]].head())
```
